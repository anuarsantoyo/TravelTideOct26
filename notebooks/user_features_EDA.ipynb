```python
import pandas as pd
import seaborn as sns
```

```python
user_features = pd.read_csv("../data_preprocessed/user_features.csv")
```

```python
user_features
```

# Civil Status Analysis

```python
ax = sns.countplot(data=user_features, x="married", hue="has_children")
ax.set_title("Civil status: married and has_children combinations")
ax.set_xlabel("Married")
ax.set_ylabel("Number of users")
ax.legend(title="Has children")
```

# Flights


```python
flights = pd.read_csv("../data_preprocessed/flights_preprocessed.csv")
```

```python
flights
```

```python
departure = pd.to_datetime(flights["departure_time"])
flights["departure_year"] = departure.dt.year
flights["departure_month"] = departure.dt.month

ax = sns.countplot(data=flights, x="departure_month", hue="departure_year", palette="deep")
ax.set_title("Flights per departure month and year")
ax.set_xlabel("Month")
ax.set_ylabel("Number of flights")
ax.legend(title="Year")
```
